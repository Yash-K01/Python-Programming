#### Decision making statements:
- This helps to control how programs work with condition.

```python
""" Syntax: if condition:
                 block of code

Operators of Comparison: == is equal to
                         != not equal to 
                         >
                         >=
                         <
                         <=
"""
num = int(input("Enter any number: "))
if num > 0:
    print("Number is Positive.")
else:
    print("Number is Negative.")
```

```python
num = int (input("Enter any number: "))
if num > 0:
    print("Number is Positive.")
elif num == 0:
    print("Number is Zero.")
else:
    print("Number is Negative.")
```

#### Logical Operators:

```python
""" AND           OR          NOT
  T  T - T     T  T - T      T - F 
  T  F - F     T  F - T      F - T 
  F  T - F     F  T - T 
  F  F - F     F  F - F
"""
# if-else ladder:-
marks = float(input("Enter your marks: "))
if marks <= 100 and marks >= 75:
    print("Gread A.")
elif marks < 75 and marks >= 60:
    print("Gread B.")
elif marks < 60 and marks >= 45:
    print("Gread C.")
elif marks < 45 and marks >= 30:
    print("Gread D.")
elif marks < 30 and marks >= 0:
    print("Fail.")
else:
    print("Invalid marks")
```

```python
# Program to check Number 3 and 7 both divisible by given num:-
num = int(input("Enter number: "))
if num % 3 == 0 and num % 7 == 0:
    print(f"{num} is divisible by both 3 & 7.")
else:
    print(f"{num} is not divisible.")
```
