#### Loops:-

```python
user_input = input("Enter some text: ")
while user_input != "exit":
    print("You are in the loop.")
    print(f"User input was {user_input}.")
    user_input = input("Enter some text: ")
print("You are outside the loop.")
```

```python
num = int(input("Enter First Number: "))
while num > 0:
    print(f"{num} is a positive number.")
    num = int(input("Enter any number: "))
print("You have entered non - positive number. Loop stopped.")
```

```python
""" Using while loop write a program to take random positive number from user. The Loop will stoped when user will enter the non-positive number. """
num = int(input("Enter first number: "))
sum = 0
while num > 0:
    print(f"{num} is a positive number.")
    sum += num
    num = int(input("Enter any number: "))
print("You have entered non-positive number. Loop stopped.")
print(f"Sum = {sum}")
```
