1. List [] :- 
- List elements are ordered means all the elements will always appear in the same position as you have added them in the list.
- List elements are indexed: Every element in the list is assigned a unique index number based on its position.
- In python we use 2 types of indexing:

a. Positive Indexing:
- It starts from the first element 0.
- Next element is the number right to the first element is 1. The right side number of 0 is 1 and so on...
- This process will continue till the last element.
["red", "blue", "pink", "white"]  -> elements = 4
  0      1        2        3      -> index = 3
- Notice the last element had a index no. 1 less than total no. of elements.
- If list is of 20 elements then last index = 19.

```python
my_list = [1, 3, "hello", True, 89.45]
print(my_list)
```

```python
print(type(my_list)) # This Identify List data type.
```

```python
""" Using index numbers we can now access the individual element from the list. """
print(my_list[0])
print(my_list[1])
```

```python
""" To know the last element index number, first need to know the total elements in the list.
- len() function gives the count of the total number of elements in a collection. 
- Suppose list has 5 elements then last index number will (n-1) => (5-1) = 4 """

print(len(my_list))
```
