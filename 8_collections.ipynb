1. List [] :- 
- List elements are ordered means all the elements will always appear in the same position as you have added them in the list.
- List elements are indexed: Every element in the list is assigned a unique index number based on its position.
- In python we use 2 types of indexing:

a. Positive Indexing:
- It starts from the first element 0.
- Next element is the number right to the first element is 1. The right side number of 0 is 1 and so on...
- This process will continue till the last element.
["red", "blue", "pink", "white"]  -> elements = 4
  0      1        2        3      -> index = 3
- Notice the last element had a index no. 1 less than total no. of elements.
- If list is of 20 elements then last index = 19.

```python
my_list = [1, 3, "hello", True, 89.45]
print(my_list)
```

```python
print(type(my_list)) # This Identify List data type.
```

```python
""" Using index numbers we can now access the individual element from the list. """
print(my_list[0])
print(my_list[1])
```

```python
""" To know the last element index number, first need to know the total elements in the list.
- len() function gives the count of the total number of elements in a collection. 
- Suppose list has 5 elements then last index number will (n-1) => (5-1) = 4 """

print(len(my_list))
```

```python
# Create a list of fruits
fruits = ["Apple", "Banana", "Mango", "Orange"]

# Print the complete list
print("Fruits:", fruits)

# Access list elements
print("First Fruit:", fruits[0])
print("Second Fruit:", fruits[1])

# Add a new fruit
fruits.append("Grapes")

# Remove a fruit
fruits.remove("Banana")

# Print updated list
print("Updated List:", fruits)

# Find total elements
print("Total Fruits:", len(fruits))
```

3. Set:-
- A Set in Python is a collection used to store multiple values. It stores only unique elements (no duplicates) and is unordered.

Think of a set as a collection where duplicate values are automatically removed.

```python
my_set = {2, 3, 4, 5, 3, 5}
print(my_set)
```

```python
# Converting a list to a set to remove duplicates
numbers = set([1, 2, 2, 3, 4, 4])
print(numbers)
```

```python
# Set Operations:-
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union - All elements
print(A | B)

# Intersection - Common elements
print(A & B)

# Difference - Elements in A but not B
print(A - B)

# Symmetric Difference - Elements not common
print(A ^ B)
```

```python

# Student Management System Using Sets

# 1. Creating Sets
python_students = {"Ram", "Rahul", "Amit", "Sneha", "Priya"}
java_students = {"Rahul", "Sneha", "Rohit", "Kiran", "Amit"}

print("Python Students:", python_students)
print("Java Students:", java_students)


# 2. Adding New Students
python_students.add("Akash")
print("\nAfter Adding Akash:", python_students)


# 3. Adding Multiple Students
python_students.update(["Neha", "Sahil", "Vijay"])
print("\nAfter Adding Multiple Students:", python_students)


# 4. Removing Students
python_students.remove("Vijay")
print("\nAfter Removing Vijay:", python_students)


# 5. Discard Element
python_students.discard("John")
print("\nAfter Discard:", python_students)


# 6. Union - Students in Both Courses
all_students = python_students.union(java_students)
print("\nAll Students:", all_students)


# 7. Intersection - Students in Both Courses
common_students = python_students.intersection(java_students)
print("\nStudents Learning Both Languages:", common_students)


# 8. Difference - Only Python Students
only_python = python_students.difference(java_students)
print("\nOnly Python Students:", only_python)


# 9. Difference - Only Java Students
only_java = java_students.difference(python_students)
print("\nOnly Java Students:", only_java)


# 10. Symmetric Difference
different_students = python_students.symmetric_difference(java_students)
print("\nStudents Learning Only One Language:", different_students)


# 11. Membership Checking
print("\nIs Ram in Python Class?", "Ram" in python_students)
print("Is Rohit in Python Class?", "Rohit" in python_students)


# 12. Subset Example
basic_students = {"Ram", "Rahul"}

print("\nIs Basic Set a Subset?",
      basic_students.issubset(python_students))


# 13. Superset Example
print("Is Python Set a Superset?",
      python_students.issuperset(basic_students))


# 14. Set Length
print("\nTotal Python Students:", len(python_students))


# 15. Loop Through Set
print("\nPython Student List:")
for student in python_students:
    print(student)


# 16. Set Comprehension
names_starting_with_A = {
    name for name in all_students if name.startswith("A")
}

print("\nNames Starting With A:", names_starting_with_A)


# 17. Copy Set
backup_students = python_students.copy()
print("\nBackup Students:", backup_students)


# 18. Pop Element
removed_student = python_students.pop()
print("\nRandomly Removed Student:", removed_student)


# 19. Clear Set
temporary_set = {"Amit", "Rahul", "Sneha"}
temporary_set.clear()

print("\nAfter Clearing:", temporary_set)


# 20. Duplicate Removal
marks = [80, 90, 80, 75, 90, 85, 75, 95]

unique_marks = set(marks)

print("\nOriginal Marks:", marks)
print("Unique Marks:", unique_marks)
```

```python
student = ("Ram", 22, "MCA", 85.5)

print(student)
print(student[0])
print(student[1])
print(student[2])
```

```python
students = {
    "Rahul" : 85,
    "Priya" : 92,
    "Amit" : 78,
    "sneha" : 88
}

# 1. Print all students and marks
for name, marks in students.items():
    print(name, marks)

# 2. Add a new student
students["Krish"] = 95

# 3. Update marks
students["Amit"] = 82

# 4. Delete a student
del students["Rahul"]

# 5. Find the student with highest marks
topper = max(students, key=students.get)
print("Topper:", topper)
print("Highest Marks:", students[topper])

# 6. Check whether a student exists
if "Priya" in students:
    print("Student found!")
else:
    print("Student not found!")

print(students)
```

```python
cart = {
    "Laptop": 50000,
    "Mouse": 500,
    "Keyboard": 1200,
    "Headphones": 2000
}

# Display all products
print("Available Products:")
for product, price in cart.items():
    print(product, ":", price)
```
